```python
import nibabel as nib
import numpy as np
import pandas as pd
import os
import os.path as osp
from glob import glob
```

```python
data_dir = '/Users/azhylka/Projects/TUMor_Data/cmaes_results'
```

```python
methods = [m for m in os.listdir(data_dir) if osp.isdir(osp.join(data_dir, m))]
methods
```

```python
def getLossOfRun(res):
    lossDir = res["lossDir"]
    minLoss = 1
    for i in range(len(lossDir)):
        for j in range(len(lossDir[i])):
            if lossDir[i][j]["lossTotal"] < minLoss:
                minLoss = lossDir[i][j]["lossTotal"]
                diceFlair = lossDir[i][j]["diceFlair"]
                diceT1c = lossDir[i][j]["diceT1c"]
                opt = lossDir[i][j]["allParams"]
                bestLossDir = lossDir[i][j]

    return minLoss, opt, bestLossDir, diceFlair, diceT1c
```

```python
subj = 'Brats2021_00014'
summary = {'Method': [], 'Dice T1c':[], 'Dice Flair': [], 'Setting': []}
for D_method in methods:
    search_dir = osp.join(data_dir, D_method, subj)
    if not osp.exists(search_dir):
        continue
    results = [r for r in os.listdir(search_dir) if r.endswith('results.npy')]
    print(results)
    for result in results:
        setting = result[len(f'sub-{subj}_'):result.find('_results.npy')]
        result = np.load(osp.join(search_dir,result), allow_pickle=True).item()
        minLoss, opt, bestLossDir, diceFlair, diceT1c = getLossOfRun(result)
        summary['Method'].append(D_method)
        summary['Setting'].append(setting)
        summary['Dice Flair'].append(diceFlair)
        summary['Dice T1c'].append(diceT1c)
```
